Explore ingested political trades, prices and macro data.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import pandas as pd
from src.db.engine import get_connection
con = get_connection(read_only=True)
```

```python
trades = con.execute('SELECT * FROM political_trades').fetch_df()
trades.describe(include='all').T.head(30)
```

```python
trades.groupby('tx_type').size().plot(kind='bar', title='Trades by type');
```

```python
prices = con.execute("SELECT * FROM prices_daily WHERE ticker='AAPL'").fetch_df()
prices.set_index('date')['adj_close'].plot(title='AAPL adjusted close');
```
